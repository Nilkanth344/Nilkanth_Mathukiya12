```python
X = df.drop(columns='alone',axis=1)
y = df['alone']
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
df.dtypes.isin(['object', 'category', 'bool'])
```

```python
encoder = LabelEncoder()
```

```python
df['sex'] = encoder.fit_transform(df['sex'])
df['embarked'] = encoder.fit_transform(df['embarked'])
df['class'] = encoder.fit_transform(df['class'])
df['who'] = encoder.fit_transform(df['who'])
df['adult_male'] = encoder.fit_transform(df['adult_male'])
df['embark_town'] = encoder.fit_transform(df['embark_town'])
df['alive'] = encoder.fit_transform(df['alive'])
df['alone'] = encoder.fit_transform(df['alone'])
df['deck'] = encoder.fit_transform(df['deck'])
```

```python
# Identify categorical columns to encode (excluding the target 'alone')
categorical_cols = X.select_dtypes(include=['object', 'category', 'bool']).columns
print(f"Categorical columns to encode: {list(categorical_cols)}")

# Apply one-hot encoding using pd.get_dummies
X_encoded = pd.get_dummies(X, columns=categorical_cols, drop_first=True, dtype=int)

print("\nX head after get_dummies:")
display(X_encoded.head())
print(f"\nShape of X after get_dummies: {X_encoded.shape}")
```

```python
from sklearn.model_selection import train_test_split
```

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
# Re-split the data with the encoded X
X_train, X_test, y_train, y_test = train_test_split(X_encoded, y, test_size=0.2, random_state=42)

print(f"X_train shape: {X_train.shape}")
print(f"y_train shape: {y_train.shape}")
print(f"X_test shape: {X_test.shape}")
print(f"y_test shape: {y_test.shape}")
```

```python
# Re-fit the model with the numerically encoded training data
model_new = RandomForestClassifier(n_estimators=50)
```

```python
model_new.fit(X_train, y_train)
```

```python
y_pred = model_new.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score, f1_score, recall_score, classification_report, confusion_matrix
```

```python
acc = accuracy_score(y_test,y_pred)
```

```python
print(f'Your accuracy is {acc * 100:.2f}%')
```

Previos Model is OverFitting and add Regularization

```python
regularized_rf = RandomForestClassifier(
    n_estimators=150,          # More trees add stability without causing overfitting
    max_depth=6,               # Hard stop on deep, overfitted tree branches
    min_samples_split=10,      # Higher number forces larger, more generalized splits
    min_samples_leaf=5,        # Prevents tiny, isolated leaf nodes
    max_features='sqrt',       # Selects a random subset of features per split
    max_samples=0.8,           # Each tree trains on a random 80% bootstrap sample
    ccp_alpha=0.01,            # Drops complex subtrees using minimal cost-complexity
    random_state=42            # Guarantees reproducible splits
)
```

```python
regularized_rf.fit(X_train, y_train)
```

```python
y_new_pred = regularized_rf.predict(X_test)
```

```python
acc = accuracy_score(y_test,y_new_pred)
```

```python
print(f'Your accuracy is: {acc * 100:.2f}%')
```

```python
f_score = f1_score(y_test,y_new_pred)
print(f'Your f1_score is: {f_score * 100:.2f}%')
```

```python
print("Classification Report:\n", classification_report(y_test, y_new_pred))
```

```python
print("\nConfusion Metrics:\n", confusion_matrix(y_test,y_new_pred))
```

```python
print("Recall Score is:", recall_score(y_test,y_new_pred))
```
